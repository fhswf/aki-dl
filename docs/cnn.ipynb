# CNN - Convolutional Neural Network

Convolutional Neural Networks (Faltungsnetzwerke) werden in erster Linie für die Bilderkennungs- und Computer Vision-Aufgaben verwendet. 

Sie bestehen zumeist aus Faltungsschicht(en), der zugehörigen Poolingschicht(en) und einem Fully Connected Layer. Der Fully Connected Layer wird ganz am Ende des Netzwerkes benutzt um eine Klassifikation oder Regression zu ermöglichen. 

Die Aufgabe der Faltungsschicht ist die Extraktion von Merkmalen in Bildern. Die ersten Faltungsschichten erkennen allgemeinere Merkmale (wie z.B. Kanten), während späteren Schichten spezifischere Merkmale des Datensatzes erkennen können. Die Pooling-Schicht wird zur Dimensionsreduktion eingesetzt. 

## Faltungsschicht

Die Faltungsschicht nutzt einen Filter mit trainierbaren Gewichten. Diese Gewichte werden für das gesamte Bild verwendet ("Parameter-Sharing"), zusätzlich kann ein BIAS zum Einsatz kommen. Der Filter wird über die Eingangsdaten in einer definierten Schrittweise bewegt. Die Eingangsdaten werden mit den zugehörigen Gewichten multipliziert und am Ende wird aus allen die Summe gebildet. Dieses Ergebnis wird in den Ausgangsdaten verwendet. Der Filter wird solange verschoben, bis die vollständigen Eingangsdaten verarbeitet wurden. Es gibt verschiedene Hyperparameter, die die Eigenschaften der Faltungsschicht beeinflussen. 
Die Hyperparameter sind häufig voneinander abhängig, weshalb diese aufeinander abgestimmt sein müssen. 

- *input_size*: Die Eingangsgröße
- *output_size*: Die Ausgangsgröße
- *kernel_size*: Die Größe des Filters.  
- *stride*: Die Schrittweite beim Verschieben des Filters. 
- *padding*: Ergänzung von zusätzlichen Schichten an den Rändern.
- *dialation*: Der Abstand der Filterelement. Normalerweise 1. Bei größeren Zahlen ist der Filter nicht mehr zusammenhängen, sondern wird mit Abstand realisiert.
- *BIAS*: Ja oder Nein

## Pooling Layer

Die Pooling Layer wird i.d.R. direkt hinter einer Faltungsschicht angewendet. Ihre Aufgabe ist das reduzieren der Dimensionen. Die Layer besitzen selbst keine Gewichte, sondern arbeiten anhand eines festgelegten Algortihmus. Dieser Algorithmus aggregiert die Eingangsdaten und gitb die reduzierten Ausgangsdaten aus. 

Beim Pooling wird ein Filter mit einer definierten Größe (kernel_size) und einer Schrittweite (stride) genutzt. Die beiden Konzepte funktionieren identisch wie bei den Faltungsschichten.  

![Abbildung eines Max-Pooling Layers](https://miro.medium.com/v2/resize:fit:1100/format:webp/0*tvWC0Uar0XD9Dcbs.png "Bild-Quelle: [3]")

Es gibt grundlegend zwei Arten von Pooling: 

- *Max-Pooling*: Beim Max Pooling wird der maximale Wert innerhalb des Filters verwendet.
- *Average-Pooling*: Beim Average Pooling wir der Durschnitte aller Werte innerhalb des Filters verwendet.  

In Pooling Schichten gehen Informationen verloren, dadurch steigt jedoch auch die Effizenz und das Risiko für Overfitting wird reduziert. 
Die Effizenz steigt dadurch, dass nachfolgende Schichten mit reduzierten Eingangsdaten auskommen und daher weniger Berechnungen notwendig sind.
