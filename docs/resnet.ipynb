# ResNet - Residual Neural Network

Residual Neural Networks wurden 2015 das erste mal vorgestellt. Sie lösen Probleme mit herkömmlichen neuronalen Netzwerken. 

Herkömmliche Netzwerke haben Probleme mit: 

- **Vanishing Gradients**: Einige Gradienten werden extrem klein, wodurch kein *Lerneffekt* entsteht, da der Gradient somit kaum einen Einfluss auf die Anpassung der Gewichte hat. 
- **Exploding Gradients**: Andersherum explodieren manche Gradienten, wodurch das Training instabil wird. 
- **Degradation in deeper neuronal networks**: Mit zunehmender Tiefe eines neuronalen Netzwerks ist das Netzwerk schlechter trainierbar und irgendwann degradiert es. Das ist problematisch, da mit zunehmenderer Tiefe komplexere und abstraktere Merkmale durch das neuronale Netzwerk wahrgenommen werden können. 

Die Residual Neural Networks begegnen dem Problem damit, dass diese *Skip Connections* einführen, wodurch dem Problem der verschwindenden Gradienten entgegengewirkt wird. 

## Funktionsweise

Ein Residual Neural Network besteht aus mehreren Residual Blocks. Grundlegend besteht ein Residual Block aus einem oder mehreren normalen Layern mit Aktivierungsfunktionen. Zusätzlich gibt es jedoch eine *Skip Connection* wodurch die Ausgabe eines Residual Blocks sich folgendermaßen bildet. 

$$H(x) = F(x) + x$$

$F(x) := \text{Funktion der Blöcke im Residual Block.}$

![Residual Neural Network](https://miro.medium.com/v2/resize:fit:1100/format:webp/1*I-sbE72ZXVIdMQnVpZdg_w.png "Aus [1], von [2]")

Das Konzept hinter Residual Neuronal Networks ist im Grundsatz, dass der Trainingsalgorithmus selbstständig "Identity-Functions/Identitäten" abbilden/finden kann. Eine Identity-Function hat dieselbe Eingabe wie Ausgabe $(G(x) = x)$. Beim Training werden die Gewichte der Layer im Residual Block auf null trainiert, wodurch nur noch die $x$ aktiv ist. 

$$ H(x) = F(x) + x \implies H(x) = x\text{, wenn} F(x) \to 0$$

Nachfolgend eine Beispiel Implementierung eines Residual Blocks:

```python
import torch.nn as nn

class ResidualBlock(nn.Module):
    def __init__(self, in_channels, out_channels, stride=1):
        super(ResidualBlock, self).__init__()
        self.conv1 = nn.Conv2d(in_channels, out_channels, kernel_size=3, stride=stride, padding=1)
        self.bn1 = nn.BatchNorm2d(out_channels)
        self.relu = nn.ReLU(inplace=True)
        self.conv2 = nn.Conv2d(out_channels, out_channels, kernel_size=3, stride=1, padding=1)
        self.bn2 = nn.BatchNorm2d(out_channels)

        self.shortcut = nn.Identity()
        if stride != 1 or in_channels != out_channels:
            self.shortcut = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, kernel_size=1, stride=stride, bias=False),
            nn.BatchNorm2d(out_channels)
        )

    def forward(self, x):
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += self.shortcut(x)  # Skip connection
        out = self.relu(out)
        return out
```
